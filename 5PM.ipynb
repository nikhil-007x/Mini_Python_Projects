```python
add = lambda a,b : print(a+b)
sub = lambda a,b : print(a-b)
div = lambda a,b : a/b if b!=0 else print("Not divisable by zero")
mul = lambda a,b : print(a*b)

ch = input("Enter Operation (+,-,*,/)")
a = int(input("Enter q : "))
b = int(input("Enter r : "))

if(ch=="+"):
    add(a,b)
elif(ch=="-"):
    sub(a,b)
elif(ch=="/"):
    sub(a,b)
elif(ch=="*"):
    mul(a,b)
else:
    print("Choose correct operation !!")
```

```python

operations = {
    "+": lambda a, b: print(a+b),
    "-": lambda a, b: print(a-b),
    "*": lambda a, b: print(a*b),
    "/": lambda a, b: print(a/b) if b != 0 else print("Not divisible by zero")
}

ch = input("Enter Operation (+,-,*,/)")
a = int(input("Enter q : "))
b = int(input("Enter r : "))

result = operations[ch](a,b)
print(result)
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fact(n-1)
n = 4
print(fact(n))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = ["A","B","C","D"]
y = [3,7,1,8]

plt.bar(x,y,color="blue")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()

data = np.random.randn(1000)
plt.hist(data,bins=20,color="orange",edgecolor='black')
plt.title("Histogram Example")
plt.show()

x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y, color='purple', marker="o")
plt.title("Scatter Plot")
plt.show()

sizes = [30,20,25,25]
labels = ["Apples","Bananas","Cherries","Dates"]
plt.pie(sizes,labels=labels,autopct="%1.1f%%",startangle=90)
plt.title("Pie Chart Example")
plt.show()

days = [1,2,3,4,5]
sleep = [7,8,6,11,7]
work = [8,7,7,8,8]
play = [1,2,3,2,1]
plt.stackplot(days,sleep,work,play , labels=["Sleep","Work","Play"])
plt.legend(loc="upper left")
plt.title("Stack Plot")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5,100)

X, Y = np.meshgrid(x, y)
Z= np.sin(np.sqrt(X ** 2+Y ** 2))
ax.plot_surface(X, Y, Z, cmap="viridis")
plt.title("3D Surface Plot")
plt.show()
```
